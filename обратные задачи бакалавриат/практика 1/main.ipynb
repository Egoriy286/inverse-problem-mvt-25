```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d

# Функция для вычисления значений f
def func(i: int, n: int) -> float:
    return 1.0

def generate_matrix_A(n, alpha, h):
    A = np.zeros((n, n))  # Создаем пустую матрицу размера n x n
    for i in range(n):
        A[i, i] = 2
        if i > 0:
            A[i, i-1] = -1
        if i < n-1:
            A[i, i+1] = -1
    return A


# Параметры
l = 1
n = 100
h = l / n
alpha = 1.5

A = generate_matrix_A(n+1, alpha, h)
f = np.array([h**2 * func(i, n) for i in range(n+1)])
g0 = 0
g1 = 1
alpha = 1.2
f[0] = g0

f[n] = f[n]* 0.5 + g1*h
A[0,:] = 0
A[0,0] = 1
A[n,n] = 1.0+alpha*h
# Решение для различных значений n
u3 = np.linalg.solve(A, f)

# Построение графиков
plt.figure(figsize=(10, 6))
plt.plot([i*h for i in range(n+1)], u3, label="u(x)", color="blue")

plt.xlabel('x')
plt.ylabel('u(x)')
plt.title('Solution')
plt.legend()
plt.grid()
print(u3)
print(A)
print(f)
plt.show()
```
